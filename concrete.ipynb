```python
import pandas as pd
import numpy as np
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error
import pickle

print("All Civil ML Libraries Imported Successfully!")
```

```python
# UCI Concrete Compressive Strength Dataset download
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/concrete/compressive/Concrete_Data.xls"
df = pd.read_excel(url)

# Columns rename for clean readability
df.columns = ['Cement', 'Slag', 'FlyAsh', 'Water', 'Superplasticizer', 
              'CoarseAggregate', 'FineAggregate', 'Age_Days', 'Strength_MPa']

print("--- First 5 Lab Records ---")
display(df.head())

print("\n--- Statistical Summary of Lab Data ---")
display(df.describe())
```

```python
# Separate Features (X) and Target (y)
X = df.drop('Strength_MPa', axis=1)
y = df['Strength_MPa']

# Split data: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Initialize and train Random Forest Regressor
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Evaluate model
predictions = model.predict(X_test)
score = r2_score(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))

print(f"Model Training Complete!")
print(f"Model Accuracy (R² Score): {score:.3f} (1.0 is perfect)")
print(f"Prediction Error (RMSE): ±{rmse:.2f} MPa")
```

```python
# Save the trained model to a file
with open('concrete_model.pkl', 'wb') as file:
    pickle.dump(model, file)

print("Model successfully saved locally as 'concrete_model.pkl'!")
```

```python
import pandas as pd
import pickle

# Load the saved model back
with open('concrete_model.pkl', 'rb') as file:
    loaded_model = pickle.load(file)

# Create a DataFrame with proper column names to match training data
columns_list = ['Cement', 'Slag', 'FlyAsh', 'Water', 'Superplasticizer', 
                'CoarseAggregate', 'FineAggregate', 'Age_Days']

custom_mix_df = pd.DataFrame([[350.0, 0.0, 0.0, 175.0, 0.0, 1020.0, 760.0, 28]], columns=columns_list)

predicted_strength = loaded_model.predict(custom_mix_df)

print(f"--- Custom Mix Prediction ---")
print(f"Estimated Compressive Strength: {predicted_strength[0]:.2f} MPa")

if predicted_strength[0] >= 30:
    print("Status: High-strength concrete mix.")
else:
    print("Status: Standard structural mix.")
```
